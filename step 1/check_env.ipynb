## سلول ۲: نسخه کتابخانه های اصلی

```python
import numpy as np
import matplotlib
import pandas as pd
import sklearn

print("NumPy version:", np.__version__)
print("Matplotlib version:", matplotlib.__version__)
print("Pandas version:", pd.__version__)
print("Scikit-learn version:", sklearn.__version__)
```

## سلول ۳: تست سلامت کتابخانه ها

```python
# NumPy: ساخت آرایه و محاسبه میانگین
arr = np.array([1, 2, 3, 4, 5])
print("NumPy array:", arr)
print("Mean:", arr.mean(), "| Std:", arr.std())

# Matplotlib: رسم یک نمودار ساده
import matplotlib.pyplot as plt
plt.bar(["A", "B", "C"], [3, 7, 5], color=["#4f46e5", "#ec4899", "#f59e0b"])
plt.title("تست سلامت Matplotlib")
plt.ylabel("مقدار")
plt.show()
```

## سلول ۴: بررسی دسترسی به دیتاست Digits

```python
from sklearn.datasets import load_digits

digits = load_digits()
print("Shape of data:", digits.data.shape)
print("Shape of target:", digits.target.shape)
print("Number of classes:", len(set(digits.target)))
print("Classes:", sorted(set(digits.target)))
print()
print("✅ دیتاست Digits با موفقیت بارگذاری شد.")
print("✅ ۱۷۹۷ نمونه تصویر ۸×۸ پیکسل از ارقام ۰ تا ۹")
```
